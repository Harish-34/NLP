```python
from nltk.util import bigrams,trigrams,ngrams
```

```python
string = 'hello the best and most beautifull thing in the world cannot be seen or even touched,they must be felt with heart'
quotes_tokens = nltk.word_tokenize(string)
quotes_tokens
```

```python
string
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams = list(nltk.bigrams(quotes_tokens))
quotes_bigrams
```

```python
quotes_trigrams = list(nltk.trigrams(quotes_tokens))
quotes_trigrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 4))
quotes_ngrams
```

```python
quotes_ngrams = list(nltk.ngrams(quotes_tokens, 6))
quotes_ngrams
```

```python
from nltk.stem import PorterStemmer
pst= PorterStemmer()
```

```python
pst.stem("affection")
```

```python
words_to_stem = ["give", "giving","given","gaved","thinking","loving","maximum"]
for words in words_to_stem:
    print(words+ " :" +pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()
```

```python
for words in words_to_stem:
    print(words+ " :" +lst.stem(words))
```

```python
from nltk.stem import SnowballStemmer
sbst = SnowballStemmer("english")
for words in words_to_stem:
    print(words+ " :" +sbst.stem(words))
```
